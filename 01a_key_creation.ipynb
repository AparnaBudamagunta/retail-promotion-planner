```python
# Create secrets in databricks to store the tokens
import requests
import json

# Your Databricks workspace URL
WORKSPACE_URL = "https://dbc-2f578a1f-757a.cloud.databricks.com/"

# Your Databricks personal access token
DATABRICKS_TOKEN = "your_pat_token" # Add stored databricks token

headers = {
    "Authorization": f"Bearer {DATABRICKS_TOKEN}",
    "Content-Type": "application/json"
}

# Step 1: Create scope
requests.post(
    f"{WORKSPACE_URL}/api/2.0/secrets/scopes/create",
    headers=headers,
    json={"scope": "my_scope"}
)

# Step 2: Add secret
requests.post(
    f"{WORKSPACE_URL}/api/2.0/secrets/secrets/put",
    headers=headers,
    json={
        "scope": "my_scope",
        "key": "groq_api_key",
        "string_value": "your_actual_groq_key"
    }
)

print("Secret created")
```

```python
# Check available Groq models
import requests, os


response = requests.get(
    "https://api.groq.com/openai/v1/models",
    headers={"Authorization": f"Bearer {os.environ['GROQ_API_KEY']}"}
)

models = response.json()
print("Available Groq models:")
for model in models["data"]:
    print(f"  → {model['id']}")
```
